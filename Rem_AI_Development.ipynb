```python

# REM AI - MEMORY SYSTEM

rem_memory = []

def remember(user_message, rem_reply):
    rem_memory.append({
        "user": user_message,
        "rem": rem_reply
    })

def show_memory():
    for item in rem_memory:
        print("You:", item["user"])
        print("Rem:", item["rem"])
        print("-------------------")

remember("Hello Rem!", "Hello! How are you?")
remember("I like anime.", "That's wonderful!")

show_memory()
```

```python

import sqlite3

# Create memory database
db = sqlite3.connect("rem_memory.db")
cursor = db.cursor()

cursor.execute("""
CREATE TABLE IF NOT EXISTS memories (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    category TEXT,
    information TEXT
)
""")

db.commit()

def save_memory(category, information):
    cursor.execute(
        "INSERT INTO memories (category, information) VALUES (?, ?)",
        (category, information)
    )
    db.commit()
    print("Rem saved this memory!")

def show_memories():
    cursor.execute("SELECT category, information FROM memories")
    memories = cursor.fetchall()

    for category, information in memories:
        print(category, ":", information)

print("Rem's long-term memory is ready!")
```

```python

# Save personal memories

save_memory("Name", "Sarwesh")
save_memory("Favourite Anime", "Re:ZERO")
save_memory("Favourite Character", "Rem")
save_memory("Interests", "Anime, Drawing, Electronics and Chess")

# Display saved memories
print("\n--- REM'S MEMORIES ---")
show_memories()
```

```python

def rem_recall():
    cursor.execute(
        "SELECT category, information FROM memories"
    )

    memories = cursor.fetchall()

    print("Rem: Let me recall what I know!")

    for category, information in memories:
        print(f"{category}: {information}")


rem_recall()
```

```python

def rem_chat_loop():
    print("Rem: Hello! I'm ready to talk.")
    print("Type 'bye' to end the conversation.\n")

    while True:
        message = input("You: ")

        if message.lower() == "bye":
            print("Rem: See you again!")
            break

        if "name" in message.lower():
            print("Rem: Your name is Sarwesh!")

        elif "anime" in message.lower():
            print("Rem: You love anime, especially Re:ZERO!")

        else:
            print("Rem: I'm listening! Tell me more.")

rem_chat_loop()
```
